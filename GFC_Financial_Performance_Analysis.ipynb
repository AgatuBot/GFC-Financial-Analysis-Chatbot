# Global Finance Corp. (GFC) Financial Performance Analysis

## Microsoft, Tesla, and Apple — 2023–2025

## 1. Project Objective

The objective of this analysis is to evaluate the financial performance of Microsoft, Tesla, and Apple from 2023 to 2025 using key financial indicators extracted from their annual 10-K filings.

The analysis focuses on Total Revenue, Net Income, Total Assets, Total Liabilities, and Operating Cash Flow. Year-over-year growth rates are calculated for Revenue and Net Income to identify financial trends and compare performance across the three companies.

The analysis is designed to provide clear, data-driven insights that can support the development of an AI-powered financial analysis chatbot for Global Finance Corp. (GFC).

## 2. Data Sources and Methodology

The financial data used in this analysis was manually extracted from the 2023, 2024, and 2025 annual 10-K filings of Microsoft, Tesla, and Apple.

All financial figures are presented in USD millions.

The following financial metrics were collected:

- Total Revenue
- Net Income
- Total Assets
- Total Liabilities
- Operating Cash Flow

The extracted data was organized in a CSV file and loaded into a Pandas DataFrame using Python. Year-over-year percentage changes were then calculated for Total Revenue and Net Income for each company.

The analysis also uses Pandas grouping and aggregation functions to compare financial performance by company and by year.

For Apple, the annual filings use the term "Total net sales" for revenue. For consistency across the analysis, this value is recorded as Total Revenue.

## 3. Data Loading

Pandas is used to load and analyze the financial dataset. The CSV file contains nine observations representing Microsoft, Tesla, and Apple across the 2023–2025 period.

The dataset is loaded into a Pandas DataFrame named `df` for further analysis.

```python
import pandas as pd
```

```python
df = pd.read_csv("financial_data.csv")
```

```python
df
```

## 4. Year-over-Year Growth Analysis

To identify changes in financial performance over time, year-over-year percentage growth was calculated for Total Revenue and Net Income.

The calculations compare each year's value with the previous year's value for the same company.

The first year in the dataset (2023) has no prior-year observation within the analysis period, so its growth rate is represented as NaN. This is expected and does not indicate missing source data.

```python
df['Revenue Growth (%)'] = df.groupby(['Company'])['Total Revenue'].pct_change() * 100
```

```python
df['Net Income Growth (%)'] = df.groupby(['Company'])['Net Income'].pct_change() * 100
```

```python
df
```

## 5. Company-Level Financial Summary

To compare the overall financial performance of Microsoft, Tesla, and Apple, the analysis calculates the average values of the key financial metrics across the 2023–2025 period.

This provides a high-level view of differences in revenue, profitability, assets, liabilities, and operating cash flow among the three companies.

```python
company_summary = df.groupby("Company")[[
    "Total Revenue",
    "Net Income",
    "Total Assets",
    "Total Liabilities",
    "Operating Cash Flow"
]].mean()

company_summary
```

## 6. Company-Level Observations

The company-level averages show clear differences in financial scale and profitability.

- **Apple** has the highest average revenue and net income across the three-year period, indicating strong overall financial scale and profitability.
- **Microsoft** has the highest average total assets and operating cash flow, reflecting a strong asset base and substantial cash generation.
- **Tesla** has substantially lower average net income and operating cash flow compared with Apple and Microsoft.
- The comparison highlights that larger revenue does not necessarily translate directly into the same level of profitability or cash generation across companies.

```python
year_summary = df.groupby("Year")[[
    "Total Revenue",
    "Net Income",
    "Total Assets",
    "Total Liabilities",
    "Operating Cash Flow"
]].sum()

year_summary
```

## 7. Year-Level Financial Observations

The combined financial results show an overall improvement in the financial position of the three companies from 2023 to 2025.

* Total revenue increased from **$691.97 billion in 2023** to **$792.71 billion in 2025**.
* Combined net income increased from **$184.33 billion in 2023** to **$217.70 billion in 2025**.
* Total assets increased from **$871.18 billion in 2023** to **$1.12 trillion in 2025**.
* Total liabilities increased from **$539.20 billion in 2023** to **$615.97 billion in 2025**.
* Operating cash flow increased from **$211.38 billion in 2023** to **$262.39 billion in 2025**.

Overall, the results indicate growth in revenue, profitability, asset base, and operating cash generation across the three companies during the analysis period.



```python
growth_summary = df.groupby("Company")[[
    "Revenue Growth (%)",
    "Net Income Growth (%)"
]].mean()

growth_summary
```

## 8. Growth Analysis Observations

The average year-over-year growth rates show significant differences in financial performance among the three companies.

- **Microsoft** recorded the strongest average growth, with approximately 15.3% revenue growth and 18.7% net income growth.
- **Apple** recorded moderate average revenue growth of approximately 4.2% and net income growth of approximately 8.1%.
- **Tesla** recorded approximately -1.0% average revenue growth and -49.2% average net income growth, indicating a significant decline in profitability during the period.
- The results suggest that Microsoft demonstrated the strongest overall growth momentum, while Tesla experienced the most significant deterioration in net income.

These growth trends provide useful indicators for comparing the financial performance and business momentum of the three companies.

## 10. Financial Performance Visualizations

Visualizations are used to highlight trends and differences in financial performance across Microsoft, Tesla, and Apple from 2023 to 2025.

The charts focus on revenue, net income, and year-over-year growth to provide a clear visual comparison of the companies.

```python
import matplotlib.pyplot as plt

for company in df["Company"].unique():
    company_data = df[df["Company"] == company]
    plt.plot(company_data["Year"], company_data["Total Revenue"], marker="o", label=company)

plt.title("Total Revenue by Company (2023–2025)")
plt.xlabel("Year")
plt.ylabel("Revenue (USD millions)")
plt.legend()
plt.grid(True)
plt.show()
```

```python
for company in df["Company"].unique():
    company_data = df[df["Company"] == company]
    plt.plot(company_data["Year"], company_data["Net Income"], marker="o", label=company)

plt.title("Net Income by Company (2023–2025)")
plt.xlabel("Year")
plt.ylabel("Net Income (USD millions)")
plt.legend()
plt.grid(True)
plt.show()
```

## 11. Visualization Observations

The revenue and net income visualizations highlight different financial trends across the three companies.

* **Microsoft** shows consistent growth in both revenue and net income from 2023 to 2025.
* **Apple** shows steady revenue growth, while net income declined slightly in 2024 before recovering strongly in 2025.
* **Tesla** shows relatively stable revenue between 2023 and 2024 followed by a decline in 2025. Its net income shows a significant decline across the period.
* The visual comparison demonstrates that revenue growth and profitability can move differently across companies.
* Overall, Microsoft demonstrates the strongest and most consistent growth trend, while Tesla shows the greatest deterioration in profitability.


## 12. Conclusion

The analysis of Microsoft, Tesla, and Apple from 2023 to 2025 shows significant differences in financial performance and growth.

Microsoft demonstrated the strongest overall growth, with consistent increases in revenue and net income throughout the period. Tesla experienced relatively stable revenue but a substantial decline in net income, indicating significant pressure on profitability.
